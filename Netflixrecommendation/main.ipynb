```python
movies['tag'] = movies['overview'] + movies['genre']
movies
```

```python
movies1 = movies.drop(columns= ['genre','overview'])
movies1.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=10000 , stop_words='english')
cv
```

```python
vector=cv.fit_transform(movies1['tag'].values.astype('U')).toarray()
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
movies1[movies1['title']=="The Godfather"].index[0]
```

```python
distance = sorted(list(enumerate(similarity[5])), reverse=True, key=lambda vector:vector[1])
# for i in distance[0:5]:
#     print(movies1.iloc[i[0]].title)

distance1 = sorted(list(enumerate(similarity[5])), reverse=True, key= lambda vector: vector[1] )
distance1
```

```python
def recommand(movies):
    index=movies1[movies1['title']==movies].index[0]
    distance = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda vector:vector[1])
    for i in distance[0:5]:
        print(movies1.iloc[i[0]].title)
```

```python
recommand("Benny & Joon")
```
